```python
import pandas as pd
from pathlib import Path

from modules.indirect_trust import build_trust_graph, compute_indirect_trust

DATA_DIR = Path("data")
RESULTS_DIR = Path("results")
RESULTS_DIR.mkdir(exist_ok=True)

rt_clean = pd.read_csv(DATA_DIR / "rt_clean.csv")

print("Dataset shape:", rt_clean.shape)
rt_clean.head()
```

```python
graph = build_trust_graph(
    rating_matrix=rt_clean,
    edge_threshold=0.50,
    min_corated=3
)

print("Number of users/nodes:", graph.number_of_nodes())
print("Number of trust edges:", graph.number_of_edges())
```

```python
target_user = 0
service_id = 0

it_score = compute_indirect_trust(
    rating_matrix=rt_clean,
    graph=graph,
    target_user=target_user,
    service_id=service_id,
    dmax=3,
    decay=0.7,
    edge_threshold=0.50
)

print("Indirect Trust Score:", it_score)
```

```python
indirect_trust_results = []

target_user = 0

for service_id in range(20):
    it = compute_indirect_trust(
        rating_matrix=rt_clean,
        graph=graph,
        target_user=target_user,
        service_id=service_id,
        dmax=3,
        decay=0.7,
        edge_threshold=0.50
    )

    indirect_trust_results.append({
        "user_id": target_user,
        "service_id": service_id,
        "indirect_trust": it
    })

indirect_trust_df = pd.DataFrame(indirect_trust_results)

indirect_trust_df.to_csv(
    RESULTS_DIR / "indirect_trust_user_0_sample.csv",
    index=False
)

indirect_trust_df
```
