```python
import pandas as pd
from pathlib import Path

from modules.aggregation import compute_final_trust, trust_decision

RESULTS_DIR = Path("results")

direct_df = pd.read_csv(RESULTS_DIR / "direct_trust_user_0_sample.csv")
indirect_df = pd.read_csv(RESULTS_DIR / "indirect_trust_user_0_sample.csv")

print("Direct trust shape:", direct_df.shape)
print("Indirect trust shape:", indirect_df.shape)

direct_df.head()
```

```python
final_trust_df = direct_df.merge(
    indirect_df,
    on=["user_id", "service_id"],
    how="inner"
)

final_trust_df["final_trust"] = final_trust_df.apply(
    lambda row: compute_final_trust(
        direct_trust=row["direct_trust"],
        indirect_trust=row["indirect_trust"],
        alpha=0.6
    ),
    axis=1
)

final_trust_df["trust_decision"] = final_trust_df["final_trust"].apply(
    lambda score: trust_decision(score, trust_threshold=0.60)
)

final_trust_df
```

```python
final_trust_df.to_csv(
    RESULTS_DIR / "final_trust_user_0_sample.csv",
    index=False
)

print("Saved:", RESULTS_DIR / "final_trust_user_0_sample.csv")
```
