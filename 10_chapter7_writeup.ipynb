```python
import pandas as pd
from pathlib import Path

DATA_DIR = Path("data")
RESULTS_DIR = Path("results")

rt_clean = pd.read_csv(DATA_DIR / "rt_clean.csv")

final_eval = pd.read_csv(RESULTS_DIR / "final_evaluation_metrics.csv")
detection = pd.read_csv(RESULTS_DIR / "detection_metrics.csv")
recommendation_summary = pd.read_csv(RESULTS_DIR / "recommendation_summary_user_0.csv")
access_summary = pd.read_csv(RESULTS_DIR / "access_decision_summary_user_0.csv")
summary_df = pd.read_csv(RESULTS_DIR / "final_result_summary.csv")

print("All Chapter 7 files loaded successfully.")
print("RT clean shape:", rt_clean.shape)
final_eval
```

```python
def get_metric(df, metric_name):
    return float(df.loc[df["Metric"] == metric_name, "Value"].iloc[0])

def get_detection_metric(df, metric_name):
    return float(df.loc[df["metric"] == metric_name, "value"].iloc[0])

mae = get_metric(final_eval, "MAE")
rmse = get_metric(final_eval, "RMSE")
coverage = get_metric(final_eval, "Coverage")
precision_n = get_metric(final_eval, "Precision@N")
recall_n = get_metric(final_eval, "Recall@N")
access_consistency = get_metric(final_eval, "Access Decision Consistency")

detection_rate = get_detection_metric(detection, "Detection Rate")
false_positive_rate = get_detection_metric(detection, "False Positive Rate")

known_malicious = int(get_detection_metric(detection, "Known Malicious Users"))
combined_flagged = int(get_detection_metric(detection, "Total Combined Flagged Users"))

services_recommended = int(recommendation_summary["services_recommended"].iloc[0])
highest_trust = float(recommendation_summary["highest_final_trust"].iloc[0])
lowest_trust = float(recommendation_summary["lowest_recommended_final_trust"].iloc[0])

print("MAE:", mae)
print("RMSE:", rmse)
print("Coverage:", coverage)
print("Precision@N:", precision_n)
print("Recall@N:", recall_n)
print("Detection Rate:", detection_rate)
print("False Positive Rate:", false_positive_rate)
print("Services Recommended:", services_recommended)
print("Highest Trust:", highest_trust)
print("Lowest Trust:", lowest_trust)
print("Access Consistency:", access_consistency)
```
